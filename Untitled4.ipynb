# analisis de los datos

## a. evolucion temporal de nacimientos.

# En esta sección analizaremos la evolucion temporal de los nacimientos y como se fue desarrollando dicha actividad en el tiempo establecido de nuestro CSV (2012 - 2024)

```python
# agrupamos por año y sumamos el total de nacimientos de cada periodo
evolucion_anual = df.groupby("año")["nacimientos"].sum()
print(evolucion_anual)
```

```python
#ordenamos de mayor a menor para ver el podio correcto
ranking_nacimientos =  evolucion_anual.sort_values(ascending=False)
print(ranking_nacimientos)
```

## b. analisis general de la caida de natalidad en la region

## Al analizar la evolución temporal de nacimientos(2012-2024), se observa un decrecimiento notable y constante a partir del periodo 2016-2024. El objetivo de este apartado es determinar si esta disminución afectó de forma generalizada a toda la region o si estuvo concentrada en los departamentos de mayor volumen de nacimiento



```python
# filtramos los años claves de la caída (2016 y 2024) y agrupamos por departamento
comparacion_depto = df[df["año"].isin([2016, 2024])].groupby(["departamento_id", "año"])["nacimientos"].sum().unstack()
# calculamos la diferencia absoluta entre ambos años
comparacion_depto["diferencia"] = comparacion_depto[2024] - comparacion_depto[2016]
# ordenamos de menor a mayorr para ver cuáles departamentos sufrieron la mayor caída
print(comparacion_depto.sort_values(by="diferencia").head(10))
```

# Nota metodológica y conceptual

Del analisis comparativo se comprende que,si bien la tendencia decreciente de la natalidad es un fenómeno que afecta a todo el territorio nacional a partir del periodo 2016-2024, el mayor impacto en términos absolutos se concentran en los departamentos de mayor tasa poblacional(teniendo al distrito ID 0.0 como el mayor). Esta correlación obedece al propio volumen poblacional de las regiones.

#Visualización

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Agrupamos los datos totales por año para todo el periodo disponible
nacimientos_por_anio = df.groupby("año")["nacimientos"].sum()

# Configuramos el grafico
plt.figure(figsize=(11, 5))
plt.plot(nacimientos_por_anio.index, nacimientos_por_anio.values, marker='o', color= "firebrick", linewidth=2.5, label="total nacimientos")

# Destacamos el año con mayor natalidad (2014)
pico_2014 = nacimientos_por_anio.loc[2014] if 2014 in nacimientos_por_anio.index else None
if pico_2014:
  plt.annotate('pico máximo (2014)', xy=(2014, pico_2014), xytext=(2014, pico_2014 + 15000), arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6), fontweight='bold', fontsize=10, ha='center')

plt.title("evolucion de la natalidad en el pais (2012-2024): del pico a la caída sostenida", fontsize=13, fontweight='bold')
plt.xlabel("año", fontsize=12)
plt.ylabel("total de nacimientos", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(nacimientos_por_anio.index) # para asegurarnos de que se vean todos los años

plt.tight_layout()
plt.show()
```
